2. "#sql"

```python
data = [
        ("Alice", "Engineering", 95000),
        ("Bob", "Marketing", 72000),
        ("Carol", "Engineering", 105000),
        ("David", "HR", 65000),
        ("Eva", "Engineering", 98000)
    ]

sch = "name string, department string, salary int"

emp_df_raw = spark.createDataFrame(data, sch)

emp_df_raw.createOrReplaceTempView("employee")
```

```python
%sql
SELECT
    department,
    COUNT(department) AS head_count,
    ROUND(AVG(salary), 2) AS avg_salary,
    MAX(salary) AS max_salary
FROM
    employee
GROUP BY department
ORDER BY avg_salary DESC;
```
